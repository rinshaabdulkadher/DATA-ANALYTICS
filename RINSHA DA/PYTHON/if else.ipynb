##  IF ELSE TASKS

```python
#1. Check whether a number is positive, negative, or zero

x=int(input("number"))
if x >0:
    print("positive")
elif x<0:
    print("negative")
else  :
    print("zero")
```

```python
#2. Check whether a number is even or odd.

x=int(input("num"))
if x% 2==0:
   print("even")
else:
    print("odd")
```

```python
#3. Find the larger of two numbers.
x=int(input("number1:"))
y=int(input("number2:"))
if x>y:
    print("x is greater")
else :
    print("y is greater")
```

```python
#4. Find the largest among three numbers.

x=int(input("no1:"))
y=int(input("no2:"))
z=int(input("no3:"))
if x>=y and x>=z:
    print("no1 is greateer")
elif y>=x and y>=z:
    print("no2 is greater")
else :
    print("no3 is greater")
```

```python
#5. Check whether a number is divisible by both 3 and 5

x=int(input("number"))
if x% 3==0 and x%  5==0:
    print("divisible by 3,5")
else:
    print("not divisible by 3,5")
```

```python
#6. Check whether a year is a leap year.


x=int(input("year"))
if x%400 == 0:
    print("leap year")
elif x% 4==0:
    print(" leap year")
elif x%100==0:
    print("leap year")
else :
    print("not leap year")
```

```python
#9. Assign a grade based on marks (A, B, C, D, Fail).
x=int(input("marks"))
if x>=90:
    print("A")
elif x>=80:
    print("B")
elif x>=70:
    print("C")
else :
    print("D")
```

```python
#14. Create a simple calculator using the user's choice of operator (+, -, *, /).


a=int(input("number"))
b=int(input("2nd number"))
operator=input("enter the operator(+,-,*,/) :")
if operator== "+":
    print("sum=",a+b)
elif operator=="-":
    print("sum=",a-b)
elif operator=="*":
    print("sum=",a*b)
elif operator=="/":
    print("sum=",a/b)

else:
    print("invalid")
```
