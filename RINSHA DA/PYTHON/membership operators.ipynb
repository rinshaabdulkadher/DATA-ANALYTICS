```python
# check whether the  first  character of string is present in the secound half of the same string

string=input("string:")

firstcharacter=string[0]

secoundhalf=string[len(string)//2:]

present=firstcharacter in secoundhalf


print("is the first character in the secound half")
print(present)   
```

```python
#check if all vowels (a,e,o,u,) are in a word
word=input("word:").lower()
vowels=["a","e","i","o","u",]
x=all(vowel in word for vowel in vowels)
print(x)    
```
