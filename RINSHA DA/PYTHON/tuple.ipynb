###   python tuple -basics task

```python
#  create a tuple of 5 fruits and print first and last elements

x=("apple","banana","cherry","kiwi","orange",)
print(x[0])
print(x[-1])
```

```python
#   find the length of the following tuple   fruits=("apple","banana","mango,"graoes")
fruits=("apple","banana","mango","grapes",)
print(len(fruits))
```

```python
#  count how many times 10 appears in the following tuple

num=(10,20,10,30,10,40,)

print(num.count(10))
```

```python
#   check whether 25 is present the following tuple

num=(10,20,30,40,50)

print(num.count(25))
```

```python
#  find the index position of 30 in the following tuple

num=(10,20,30,40,50)

print(num.index(30))
```

```python
#  add 60 to the following tuple

num=(10,20,30,40,50,)
x=list(num)
x.append(60)
num=tuple (x)
print(x)
```

```python
#  remove 30 from the following tuple

num=(10,20,30,40,50,)
x=list(num)
x.remove(30)
num=tuple(x)
print(x)
```

```python
#  join the following two tuples into a single tuple

x=(1,2,3)
y=(4,5,6)
print(x + y)
```

```python
##  replace 20 with 200 in the following tuple

num=(10,20,30,40,50)
y=list(num)
y[1]=200
x=tuple(y)
print(x)
```

```python
#add a new elements to a tuple and remove an existing element from it

num=(10,20,30,40,50,)
y=list(num)
y.append(60)
y.remove(20)

x=tuple(y)
print(x)
```
