```python
#1. Find the common elements between the following two sets:
students_python = {"Anu", "Rahul", "Meera", "Arun"}
students_java = {"Rahul", "Meera", "Vishnu", "Neha"}

x=students_python.intersection(students_java)

print(x)
```

```python
#2. Find the students who are enrolled in Python but not Java:
python = {"Anu", "Rahul", "Meera", "Arun"}
java = {"Rahul", "Meera", "Vishnu"}
x=python.difference(java)

print(x)
```

```python
#3. Combine the following two sets and remove duplicate values:
set1 = {10, 20, 30, 40}
set2 = {30, 40, 50, 60}

x=set1.union(set2)
print(x)
```

```python
#4. Find the elements that are unique to each set:
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}
x=set1.symmetric_difference(set2)
print(x)
```

```python
#5. Check whether set1 is a subset of set2 :
set1= {10, 20, 30}
set2= {10, 20, 30, 40, 50}

print(set1.issubset(set2))
```

```python
#6. Check whether the following two sets are disjoint:
set1 = {1, 2, 3}
set2 = {4, 5, 6}

print(set1.isdisjoint(set2))
```

```python
#7. Find the elements that are present in both sets:
set1 = {"Apple", "Banana", "Mango", "Orange"}
set2 = {"Banana", "Orange", "Grapes"}

x=set1.intersection(set2)
print(x)
```

```python
#8. Add "Python" to the following set and remove "Java" :
courses = {"C", "Java", "SQL"}

courses.add("python")
courses.remove("Java")
print(courses)
```

```python
#9. Remove duplicate values from the following list using a set:
marks = [80, 90, 75, 80, 95, 90, 85]

x=set(marks)
print(x)
```

```python
#10. Find the union, intersection, and difference of the following two sets:
A = {10, 20, 30, 40, 50}
B = {30, 40, 50, 60, 70}

x=A.union(B)
y=A.intersection(B)
z=A.difference(B)

print(x)
print(y)
print(z)
```
