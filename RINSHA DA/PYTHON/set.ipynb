```python
#6. Find the union of the following sets:
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

set2.union(set1)
```

```python
#7. Find the intersection of the following sets:
set1 = {10, 20, 30, 40}
set2 = {30, 40, 50, 60}

set3=set2.intersection(set1)
print(set3)
```

```python
#8. Find the elements that are present in the first set but not in the second set:
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7}
set3=set2.difference(set1)
print(set3)
```
