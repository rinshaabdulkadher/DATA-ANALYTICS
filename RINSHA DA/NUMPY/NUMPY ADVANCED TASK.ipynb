```python
#10. Given the array [10, 20, 30, 40, 50] , find the total, average, median,standard deviation, variance, smallest value, and largest value.


t=([10,20,30,40,50])
print(np.sum(t))
print(np.mean(t))
print(np.median(t))
print(np.std(t))
print(np.var(t))
print(np.min(t))
print(np.max(t))
```

```python
#11. Given the array [10, 20, 5, 40, 15] , find the positions of the smallest and largest elements.
h=np.array([10,20,5,40,15])
print(np.argmin(h))
print(np.argmax(h))
```

```python
#12. Given the array:
#[[2, 4, 6],
#[8, 10, 12]]
#Find the sum of the values in each row and the sum of the values in each column.


l=np.array([[2,4,6],
            [8,10,12]])
print("colum wise:",np.sum(l, axis=0))
print("row wise:",np.sum(l, axis=1))
```

```python
#13. Create an array containing [10, 20, 30, 40] . Create a separate copy and a view of the array. Modify the arrays and observe the effect 
#on the original array.

hu=np.array([10,20,30,40])
a=hu.copy()
a[0]=100
print(hu)
print(a)
b=hu.view()
b[0]=100

print(b)

```

```python
#15. Given the array:
#[[1, 2, 3],
#[4, 5, 6]] Print every element one by one, then convert the array into a one-dimensional array.

op=np.array([[1,2,3,],
             [4,5,6]])
for i in np.nditer(op):
    print(i)
print(op.flatten())
```
