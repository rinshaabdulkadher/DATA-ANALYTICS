```python
#7. Given the following array of marks,
marks = np.array([45, 78, 92, 56, 34, 88, 67, 95, 41, 73])
#a. Create a filtered array of marks greater than 60.

a = marks[marks > 60]
print(a)
```

```python
#7b. Create a filtered array of marks between 50 and 80.



b=marks[(marks >= 50) & (marks <= 80)]
print(b)
```

```python
#7c. Create a filtered array of marks greater than or equal to 90.

c=marks[marks >= 90]
print(c)
```
