#                 NumPy Array Operations task


```python
import numpy as np
```

```python
#1. Given the array:
#a = np.array([10, 20, 30, 40, 50, 60])
#Print the elements from index 1 to index 4.


a =np.array ([10, 20, 30, 40, 50, 60])
print(a[1:5])


              
```

```python
# 2. Given the array:
#a = np.array([1, 2, 3, 4], dtype=np.int32)
#Find and print the data type, item size, and total memory used by the
#array.
a = np.array([1, 2, 3, 4], dtype=np.int32)

print(a.dtype)
print(a.itemsize)
print(a.nbytes)
```

```python
#3. Given the array:
#a = np.array([[10, 20, 30], [40, 50, 60]])
#Print its shape, then reshape it into 3 rows and 2 columns.


a = np.array([[10, 20, 30], [40, 50, 60]])
print(a.shape)
print(a.reshape(3,2))
```

```python
#4. Given the array:
#a = np.array([[1, 2], [3, 4]])

#Create a view of the array, change the first element of the view to 100 , and
#print both the original array and the view. Check the base attribute.

a = np.array([[1, 2], [3, 4]])
b=a.view()
b[0]=100
print(a)
print(b)
print(b.base)
```
