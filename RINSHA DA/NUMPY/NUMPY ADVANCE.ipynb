 #       ARRAY COPY & VIEW

```python
fg=np.array([10,20,30,40,50])
fh=fg.copy()
fg[3]=89
print(fh)
print(fg)
```

```python
fg=np.array([10,20,30,40,50])
fj=fg.view()
fg[4]=200
print(fg)
print(fj)
```

```python
a = np.array([1, 2, 3, 4])
b = a.copy()
c = a.view()
print(b.base)
print(c.base)
```
